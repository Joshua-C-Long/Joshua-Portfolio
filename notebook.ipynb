You're working for a company that sells motorcycle parts, and they've asked for some help in analyzing their sales data!

They operate three warehouses in the area, selling both retail and wholesale. They offer a variety of parts and accept credit cards, cash, and bank transfer as payment methods. However, each payment type incurs a different fee.

The board of directors wants to gain a better understanding of wholesale revenue by product line, and how this varies month-to-month and across warehouses. You have been tasked with calculating net revenue for each product line and grouping results by month and warehouse. The results should be filtered so that only `"Wholesale"` orders are included.

They have provided you with access to their database, which contains the following table called `sales`:

## Sales
| Column | Data type | Description |
|--------|-----------|-------------|
| `order_number` | `VARCHAR` | Unique order number. |
| `date` | `DATE` | Date of the order, from June to August 2021. |
| `warehouse` | `VARCHAR` | The warehouse that the order was made from&mdash; `North`, `Central`, or `West`. |
| `client_type` | `VARCHAR` | Whether the order was `Retail` or `Wholesale`. |
| `product_line` | `VARCHAR` | Type of product ordered. |
| `quantity` | `INT` | Number of products ordered. | 
| `unit_price` | `FLOAT` | Price per product (dollars). |
| `total` | `FLOAT` | Total price of the order (dollars). |
| `payment` | `VARCHAR` | Payment method&mdash;`Credit card`, `Transfer`, or `Cash`. |
| `payment_fee` | `FLOAT` | Percentage of `total` charged as a result of the `payment` method. |


Your query output should be presented in the following format:

| `product_line` | `month` | `warehouse` |	`net_revenue` |
|----------------|-----------|----------------------------|--------------|
| product_one | --- | --- | --- |
| product_one | --- | --- | --- |
| product_one | --- | --- | --- |
| product_one | --- | --- | --- |
| product_one | --- | --- | --- |
| product_one | --- | --- | --- |
| product_two | --- | --- | --- |
| ... | ... | ... | ... |

```python
-- Start coding here
-- Start coding here
SELECT product_line,
    CASE WHEN EXTRACT('month' from date) = 6 THEN 'June'
        WHEN EXTRACT('month' from date) = 7 THEN 'July'
        WHEN EXTRACT('month' from date) = 8 THEN 'August'
    END as month,
    warehouse,
	SUM(total) - SUM(payment_fee) AS net_revenue
FROM sales
WHERE client_type = 'Wholesale'
GROUP BY product_line, warehouse, month
ORDER BY product_line, month, net_revenue DESC
```

### How did net wholesale revenue change from June to July to August?

```python
SELECT
    EXTRACT('month' FROM date) AS month,
    SUM(total) - SUM(payment_fee) AS net_revenue
FROM sales
WHERE client_type = 'Wholesale'
GROUP BY month
ORDER BY month ASC;
```

### Findings
Net wholesale revenue remained nearly unchanged from June ($49,099.18) to July ($49,085.35). In August, revenue increased substantially to $61,455.55, representing an increase of approximately 25% compared with July. This indicates a significant improvement in wholesale sales performance during August and suggests further analysis could identify which products or warehouses contributed most to the increase.

### What caused the August revenue increase—which product lines contributed most to the growth?

```python
SELECT
    product_line,

    SUM(CASE
        WHEN EXTRACT('month' FROM date) = 7
        THEN total - payment_fee
        ELSE 0
    END) AS july_revenue,

    SUM(CASE
        WHEN EXTRACT('month' FROM date) = 8
        THEN total - payment_fee
        ELSE 0
    END) AS august_revenue

FROM sales

WHERE client_type = 'Wholesale'
    AND EXTRACT('month' FROM date) IN (7, 8)

GROUP BY product_line

ORDER BY product_line;
```

### Findings
The increase in August net wholesale revenue was driven primarily by strong growth in the Engine and Frame & body product lines. Engine revenue increased from about $2,834 in July to $11,853 in August, while Frame & body increased from about $9,290 to $17,387. Electrical system also increased modestly, while Braking system, Miscellaneous, and Suspension & traction declined. This suggests that the overall August revenue increase was concentrated in a few high-growth product categories rather than across all product lines.

### How did payment fees affect total wholesale revenue?

```python
SELECT
    payment,
    SUM(payment_fee) AS total_payment_fees,
    SUM(total) AS gross_revenue,
    SUM(total) - SUM(payment_fee) AS net_revenue
FROM sales
WHERE client_type = 'Wholesale'
GROUP BY payment
ORDER BY total_payment_fees DESC;
```

### Findings
Wholesale transactions generated $159,642.33 in gross revenue and incurred only $2.25 in payment fees, resulting in $159,640.08 in net revenue. Since wholesale orders used bank transfers, payment fees had a minimal impact on overall wholesale revenue.
